```python
from coffea.nanoevents import NanoEventsFactory, BaseSchema
from L1LLPJetTagger.processor import TreeExplorer
from L1LLPJetTagger.plotting import plot_hist
from hist import Hist
import uproot
import awkward as ak
import matplotlib.pyplot as plt

# Load file
filename = "../../jetTuple_extended_5.root"
file = uproot.open(filename)

# List keys in the file (e.g., TTrees, histograms, etc.)
file.keys()
```

```python
events = NanoEventsFactory.from_root(
    f"{filename}:outnano/Jets", schemaclass=BaseSchema
).events()

processor_instance = TreeExplorer()
output = processor_instance.process(events)
```

```python
# just some checks
print("Check output: ", output.keys())
print(output["unknown"].keys())

print("Check num entries output: ", output["unknown"]["entries"])

print("Check num entries events: ", len(events))

print(output["unknown"]["jet_pt"])
```

```python
pt_hist = Hist.new.Reg(50, 0, 500, name="pt", label="Jet $p_T$ [GeV]").Double()

pt_hist.fill(pt=output["unknown"]["jet_pt"]);
```

```python
fig, ax = plot_hist(
    pt_hist, title="Jet $p_T$ Distribution", xlabel="Jet $p_T$ [GeV]", ylabel="Jets"
)
# ax.set_yscale("log")
plt.show()
# fig.savefig("jet_pt.png", dpi=300)
```
